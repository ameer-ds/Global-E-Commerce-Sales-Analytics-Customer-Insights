```python
data["Date"] = pd.to_datetime(data["Date"], format="%d-%m-%Y")
```

```python
data.info()
```

```python
data["year"]= data["Date"].dt.year
```

```python
data["Quarter"] = data["Date"].dt.quarter
```

```python
data["Month_No"] = data["Date"].dt.month
data["Month"] = data["Date"].dt.month_name()
```

```python
data["Weekday"] = data["Date"].dt.strftime("%A")
```

```python
data.head()
```

```python
data["Revenue_Per_Unit"] = data["Total_Revenue"] / data["Quantity"]
```

```python
data["Delivery_Status"] = np.where(
    data["Delivery_Time_Days"] <= 5,
    "Fast Delivery",
    "Slow Delivery"
)
```

```python
data["Age_Group"] = pd.cut(
    data["Customer_Age"],
    bins=[18,25,35,45,55,65,100],
    labels=["18-25","26-35","36-45","46-55","56-65","65+"],
    include_lowest=True
)
```

```python
data.to_csv("Global_Ecommerce_Sales.csv",index=False)
```
